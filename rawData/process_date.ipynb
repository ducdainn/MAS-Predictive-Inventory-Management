```python
import pandas as pd
from datetime import datetime
```

```python
df_excel = pd.read_excel('Data_FSKU_updated (1).csv (1).xlsx')
df_excel.shape
```

```python
df = df_excel

# Hàm chuyển đổi định dạng ngày
def convert_date_format(date_str):
    """
    Chuyển đổi ngày từ nhiều định dạng khác nhau sang yyyy-mm-dd
    """
    date_str = str(date_str).strip()
    
    # Nếu đã có định dạng yyyy-dd-mm HH:MM:SS, parse theo định dạng này
    if len(date_str) > 10 and date_str[4] == '-' and date_str[7] == '-':
        try:
            # Parse theo định dạng YYYY-DD-MM HH:MM:SS
            date_obj = datetime.strptime(date_str, '%Y-%d-%m %H:%M:%S')
            # Chuyển sang định dạng yyyy-dd-mm
            return date_obj.strftime('%Y-%d-%m')
        except ValueError:
            # Nếu parse lỗi, thông báo lỗi
            print(f"Lỗi khi chuyển đổi ngày: {date_str}")
            return date_str
    
    # Chuyển đổi định dạng dd/mm/yy sang yyyy-mm-dd
    try:
        # Parse định dạng dd/mm/yy
        date_obj = datetime.strptime(date_str, '%d/%m/%y')
        # Chuyển sang định dạng yyyy-mm-dd
        return date_obj.strftime('%Y-%m-%d')
    except ValueError:
        # Nếu không parse được, trả về giá trị gốc
        return date_str

# Áp dụng hàm chuyển đổi cho cột 'Ngày'
df['Ngày'] = df['Ngày'].apply(convert_date_format)

# Chuyển cột 'Ngày' thành datetime để sắp xếp
df['Ngày_datetime'] = pd.to_datetime(df['Ngày'])

# # Xóa cột phụ trợ
df = df.drop('Ngày_datetime', axis=1)

# # Reset index
df = df.reset_index(drop=True)

# Phân tích theo năm
print("\n--- PHÂN TÍCH THEO NĂM ---")
df['Năm'] = pd.to_datetime(df['Ngày']).dt.year
year_counts = df['Năm'].value_counts().sort_index()
for year, count in year_counts.items():
    print(f"Năm {year}: {count} ngày")

# Phân tích theo tháng của từng năm
print("\n--- PHÂN TÍCH THEO THÁNG/NĂM ---")
df['Tháng'] = pd.to_datetime(df['Ngày']).dt.month
month_year = df.groupby(['Năm', 'Tháng']).size()
print(month_year)

# Lưu kết quả đã sắp xếp
df.to_csv('Data_FSKU_cleanedDate.csv', index=False, encoding='utf-8-sig')

# Kiểm tra dữ liệu năm 2025
print("\n--- DỮ LIỆU NĂM 2025 ---")
df_2025 = df[df['Năm'] == 2025]
if len(df_2025) > 0:
    print(f"Có {len(df_2025)} ngày trong năm 2025")
    print(f"Từ {df_2025['Ngày'].iloc[0]} đến {df_2025['Ngày'].iloc[-1]}")
    print("\nCác tháng trong năm 2025:")
    print(df_2025['Tháng'].value_counts().sort_index())
else:
    print("Không có dữ liệu năm 2025")
```
